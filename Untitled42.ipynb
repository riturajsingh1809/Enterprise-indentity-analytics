```python
import pandas as pd
import numpy as np

# ------------------------------------------------------------
# Load Cleaned Dataset
# ------------------------------------------------------------

file_name = "Cleaned_Enterprise_Identity.csv"

df = pd.read_csv(file_name)

print("Dataset Loaded Successfully")
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])


# ------------------------------------------------------------
# Basic Information
# ------------------------------------------------------------

print("\n" + "=" * 60)
print("DATASET OVERVIEW")
print("=" * 60)

print("\nColumns:")
print(df.columns.tolist())

print("\nFirst 10 Records:")
display(df.head(10))

print("\nDataset Information:")
display(
    pd.DataFrame({
        "Column": df.columns,
        "Data Type": df.dtypes.astype(str).values,
        "Missing Values": df.isnull().sum().values,
        "Unique Values": df.nunique().values
    })
)


# ------------------------------------------------------------
# Data Type Conversion
# ------------------------------------------------------------

df["membership_start"] = pd.to_datetime(
    df["membership_start"],
    errors="coerce"
)

df["event_date"] = pd.to_datetime(
    df["event_date"],
    errors="coerce"
)

df["activity_count"] = pd.to_numeric(
    df["activity_count"],
    errors="coerce"
)

df["activity_value"] = pd.to_numeric(
    df["activity_value"],
    errors="coerce"
)


# ------------------------------------------------------------
# Missing Value Analysis
# ------------------------------------------------------------

print("\n" + "=" * 60)
print("MISSING VALUE ANALYSIS")
print("=" * 60)

missing_report = pd.DataFrame({
    "Column": df.columns,
    "Missing Values": df.isnull().sum().values,
    "Missing Percentage": (
        df.isnull().mean().values * 100
    ).round(2)
})

display(missing_report)


# ------------------------------------------------------------
# Duplicate Analysis
# ------------------------------------------------------------

print("\n" + "=" * 60)
print("DUPLICATE ANALYSIS")
print("=" * 60)

duplicate_count = df.duplicated().sum()

print("Duplicate Records:", duplicate_count)


# ------------------------------------------------------------
# Basic Statistics
# ------------------------------------------------------------

print("\n" + "=" * 60)
print("NUMERICAL STATISTICS")
print("=" * 60)

numeric_columns = [
    "activity_count",
    "activity_value"
]

statistics = df[numeric_columns].describe().T

statistics["median"] = df[numeric_columns].median()
statistics["skewness"] = df[numeric_columns].skew()

display(statistics)


# ------------------------------------------------------------
# Overall KPIs
# ------------------------------------------------------------

print("\n" + "=" * 60)
print("OVERALL KPIs")
print("=" * 60)

kpi = pd.DataFrame({
    "Metric": [
        "Total Records",
        "Unique Users",
        "Total Organizations",
        "Total Roles",
        "Total Activity",
        "Average Activity",
        "Total Activity Value",
        "Average Activity Value"
    ],
    "Value": [
        len(df),
        df["user_id"].nunique(),
        df["org_name"].nunique(),
        df["role"].nunique(),
        df["activity_count"].sum(),
        df["activity_count"].mean(),
        df["activity_value"].sum(),
        df["activity_value"].mean()
    ]
})

display(kpi)


# ------------------------------------------------------------
# Organization Analysis
# ------------------------------------------------------------

print("\n" + "=" * 60)
print("ORGANIZATION ANALYSIS")
print("=" * 60)

organization_analysis = (
    df.groupby("org_name")
    .agg(
        Records=("event_id", "count"),
        Unique_Users=("user_id", "nunique"),
        Total_Activity=("activity_count", "sum"),
        Average_Activity=("activity_count", "mean"),
        Total_Value=("activity_value", "sum"),
        Average_Value=("activity_value", "mean")
    )
    .reset_index()
    .sort_values("Total_Value", ascending=False)
)

display(organization_analysis)


# ------------------------------------------------------------
# Role Analysis
# ------------------------------------------------------------

print("\n" + "=" * 60)
print("ROLE ANALYSIS")
print("=" * 60)

role_analysis = (
    df.groupby("role")
    .agg(
        Records=("event_id", "count"),
        Unique_Users=("user_id", "nunique"),
        Total_Activity=("activity_count", "sum"),
        Average_Activity=("activity_count", "mean"),
        Total_Value=("activity_value", "sum"),
        Average_Value=("activity_value", "mean")
    )
    .reset_index()
    .sort_values("Total_Value", ascending=False)
)

display(role_analysis)


# ------------------------------------------------------------
# User Status Analysis
# ------------------------------------------------------------

print("\n" + "=" * 60)
print("USER STATUS ANALYSIS")
print("=" * 60)

status_analysis = (
    df.groupby("user_status")
    .agg(
        Records=("event_id", "count"),
        Unique_Users=("user_id", "nunique"),
        Total_Activity=("activity_count", "sum"),
        Average_Activity=("activity_count", "mean"),
        Total_Value=("activity_value", "sum"),
        Average_Value=("activity_value", "mean")
    )
    .reset_index()
    .sort_values("Records", ascending=False)
)

display(status_analysis)


# ------------------------------------------------------------
# Active Users
# ------------------------------------------------------------

print("\n" + "=" * 60)
print("ACTIVE USER ANALYSIS")
print("=" * 60)

active_users = df[df["user_status"] == "Active"]

print("Active Records:", len(active_users))
print("Active Unique Users:", active_users["user_id"].nunique())
print("Active Activity:", active_users["activity_count"].sum())
print("Active Activity Value:", active_users["activity_value"].sum())


# ------------------------------------------------------------
# Deprovisioned Users
# ------------------------------------------------------------

print("\n" + "=" * 60)
print("DEPROVISIONED USER ANALYSIS")
print("=" * 60)

deprovisioned_users = df[
    df["user_status"] == "Deprovisioned"
]

print(
    "Deprovisioned Records:",
    len(deprovisioned_users)
)

print(
    "Deprovisioned Unique Users:",
    deprovisioned_users["user_id"].nunique()
)

print(
    "Deprovisioned Activity:",
    deprovisioned_users["activity_count"].sum()
)

print(
    "Deprovisioned Activity Value:",
    deprovisioned_users["activity_value"].sum()
)


# ------------------------------------------------------------
# Deleted Users
# ------------------------------------------------------------

print("\n" + "=" * 60)
print("DELETED USER ANALYSIS")
print("=" * 60)

deleted_users = df[
    df["user_status"] == "Deleted"
]

print(
    "Deleted Records:",
    len(deleted_users)
)

print(
    "Deleted Unique Users:",
    deleted_users["user_id"].nunique()
)

print(
    "Deleted Activity:",
    deleted_users["activity_count"].sum()
)

print(
    "Deleted Activity Value:",
    deleted_users["activity_value"].sum()
)


# ------------------------------------------------------------
# Organization + Role Analysis
# ------------------------------------------------------------

print("\n" + "=" * 60)
print("ORGANIZATION AND ROLE ANALYSIS")
print("=" * 60)

organization_role_analysis = (
    df.groupby(["org_name", "role"])
    .agg(
        Records=("event_id", "count"),
        Unique_Users=("user_id", "nunique"),
        Total_Activity=("activity_count", "sum"),
        Total_Value=("activity_value", "sum")
    )
    .reset_index()
    .sort_values("Total_Value", ascending=False)
)

display(organization_role_analysis)


# ------------------------------------------------------------
# User Organization Movement
# ------------------------------------------------------------

print("\n" + "=" * 60)
print("ORGANIZATION MOVEMENT ANALYSIS")
print("=" * 60)

user_movement = (
    df.groupby("user_id")
    .agg(
        Organization_Count=("org_name", "nunique"),
        Organizations=(
            "org_name",
            lambda x: " -> ".join(
                pd.unique(x.dropna())
            )
        ),
        Record_Count=("event_id", "count"),
        Total_Activity=("activity_count", "sum"),
        Total_Value=("activity_value", "sum")
    )
    .reset_index()
)

moved_users = user_movement[
    user_movement["Organization_Count"] > 1
].copy()

print(
    "Users with Organization Movement:",
    len(moved_users)
)

display(
    moved_users.sort_values(
        "Organization_Count",
        ascending=False
    )
)


# ------------------------------------------------------------
# User Role Movement
# ------------------------------------------------------------

print("\n" + "=" * 60)
print("ROLE MOVEMENT ANALYSIS")
print("=" * 60)

user_role_movement = (
    df.groupby("user_id")
    .agg(
        Role_Count=("role", "nunique"),
        Roles=(
            "role",
            lambda x: " -> ".join(
                pd.unique(x.dropna())
            )
        )
    )
    .reset_index()
)

role_changed_users = user_role_movement[
    user_role_movement["Role_Count"] > 1
].copy()

print(
    "Users with Role Changes:",
    len(role_changed_users)
)

display(role_changed_users)


# ------------------------------------------------------------
# Historical Membership Analysis
# ------------------------------------------------------------

print("\n" + "=" * 60)
print("HISTORICAL MEMBERSHIP ANALYSIS")
print("=" * 60)

historical_df = df[
    df["membership_start"].notna()
    &
    df["event_date"].notna()
].copy()

historical_df["membership_days"] = (
    historical_df["event_date"]
    -
    historical_df["membership_start"]
).dt.days

historical_df.loc[
    historical_df["membership_days"] < 0,
    "membership_days"
] = np.nan

print(
    "Average Membership Days:",
    round(
        historical_df["membership_days"].mean(),
        2
    )
)

print(
    "Median Membership Days:",
    round(
        historical_df["membership_days"].median(),
        2
    )
)

print(
    "Minimum Membership Days:",
    historical_df["membership_days"].min()
)

print(
    "Maximum Membership Days:",
    historical_df["membership_days"].max()
)


# ------------------------------------------------------------
# Monthly Analysis
# ------------------------------------------------------------

print("\n" + "=" * 60)
print("MONTHLY ANALYSIS")
print("=" * 60)

monthly_analysis = (
    df.dropna(subset=["event_date"])
    .assign(
        Year=lambda x: x["event_date"].dt.year,
        Month=lambda x: x["event_date"].dt.month,
        Month_Name=lambda x: x["event_date"].dt.strftime("%B")
    )
    .groupby(
        ["Year", "Month", "Month_Name"]
    )
    .agg(
        Records=("event_id", "count"),
        Unique_Users=("user_id", "nunique"),
        Total_Activity=("activity_count", "sum"),
        Total_Value=("activity_value", "sum")
    )
    .reset_index()
    .sort_values(["Year", "Month"])
)

display(monthly_analysis)


# ------------------------------------------------------------
# Analytics Eligible Data
# ------------------------------------------------------------

print("\n" + "=" * 60)
print("ANALYTICS ELIGIBILITY")
print("=" * 60)

analytics_data = df[
    df["user_status"] == "Active"
].copy()

print(
    "Records Included in Analytics:",
    len(analytics_data)
)

print(
    "Users Included in Analytics:",
    analytics_data["user_id"].nunique()
)

print(
    "Activity Included in Analytics:",
    analytics_data["activity_count"].sum()
)

print(
    "Activity Value Included in Analytics:",
    analytics_data["activity_value"].sum()
)


# ------------------------------------------------------------
# Data Quality Summary
# ------------------------------------------------------------

print("\n" + "=" * 60)
print("FINAL DATA QUALITY SUMMARY")
print("=" * 60)

quality_summary = pd.DataFrame({
    "Metric": [
        "Total Records",
        "Total Columns",
        "Unique Users",
        "Unique Organizations",
        "Unique Roles",
        "Duplicate Records",
        "Missing Values",
        "Active Records",
        "Deprovisioned Records",
        "Deleted Records",
        "Analytics Eligible Records"
    ],
    "Value": [
        len(df),
        len(df.columns),
        df["user_id"].nunique(),
        df["org_name"].nunique(),
        df["role"].nunique(),
        df.duplicated().sum(),
        df.isnull().sum().sum(),
        (df["user_status"] == "Active").sum(),
        (df["user_status"] == "Deprovisioned").sum(),
        (df["user_status"] == "Deleted").sum(),
        len(analytics_data)
    ]
})

display(quality_summary)


# ------------------------------------------------------------
# Save Analysis Report
# ------------------------------------------------------------

output_file = "Enterprise_Identity_Python_Analysis.xlsx"

with pd.ExcelWriter(
    output_file,
    engine="openpyxl"
) as writer:

    df.to_excel(
        writer,
        sheet_name="Cleaned_Data",
        index=False
    )

    kpi.to_excel(
        writer,
        sheet_name="KPIs",
        index=False
    )

    missing_report.to_excel(
        writer,
        sheet_name="Missing_Values",
        index=False
    )

    organization_analysis.to_excel(
        writer,
        sheet_name="Organizations",
        index=False
    )

    role_analysis.to_excel(
        writer,
        sheet_name="Roles",
        index=False
    )

    status_analysis.to_excel(
        writer,
        sheet_name="User_Status",
        index=False
    )

    organization_role_analysis.to_excel(
        writer,
        sheet_name="Org_Role",
        index=False
    )

    moved_users.to_excel(
        writer,
        sheet_name="Org_Movement",
        index=False
    )

    role_changed_users.to_excel(
        writer,
        sheet_name="Role_Movement",
        index=False
    )

    monthly_analysis.to_excel(
        writer,
        sheet_name="Monthly_Analysis",
        index=False
    )

    quality_summary.to_excel(
        writer,
        sheet_name="Data_Quality",
        index=False
    )

print("\n" + "=" * 60)
print("ANALYSIS COMPLETED SUCCESSFULLY")
print("=" * 60)

print(
    "\nAnalysis report saved as:",
    output_file
)

print("\nNext step: Tableau Dashboard")
```
